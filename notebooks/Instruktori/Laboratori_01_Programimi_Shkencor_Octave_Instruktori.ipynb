## 2. Variablat, kushtet, ciklet dhe funksionet

MATLAB/Octave përdor indeksim që fillon nga 1. Pikëpresja `;` ndalon afishimin automatik të një rezultati. Operatorët `.*`, `./` dhe `.^` veprojnë element më element.


````octave
masa_kg = 2.5;
shpejtesia_m_s = 3.2;
energjia_J = 0.5 * masa_kg * shpejtesia_m_s^2;
fprintf('Energjia kinetike: %.3f J\n', energjia_J);

temperatura_K = 295.0;
if temperatura_K < 273.15
    gjendja = 'nen piken e ngrirjes se ujit';
elseif temperatura_K < 373.15
    gjendja = 'ndermjet pikes se ngrirjes dhe vlimit';
else
    gjendja = 'mbi piken e vlimit te ujit';
end
disp(gjendja);

for v_m_s = [0, 1, 2, 3]
    fprintf('v=%.1f m/s, E_k=%.2f J\n', v_m_s, 0.5*masa_kg*v_m_s^2);
end

n = 0;
while n < 3
    fprintf('Hapi %d\n', n);
    n = n + 1;
end
````

> **Gabim i zakonshëm:** MATLAB/Octave nuk lejon një funksion lokal në çdo pozicion në versionet e vjetra. Në skriptin `.m`, funksionet janë vendosur në fund. Në notebook përdoren formula të drejtpërdrejta dhe funksione anonime kur është e përshtatshme.


````octave
energjia_kinetike = @(m_kg, v_m_s) 0.5 .* m_kg .* v_m_s.^2;
fprintf('E_k = %.3f J\n', energjia_kinetike(2.5, 3.2));
assert(abs(energjia_kinetike(2.0, 0.0)) < 1e-14);
````

## 3. Vektorët dhe matricat

| Veprimi | Python/NumPy | MATLAB/Octave |
|---|---|---|
| Vektor i rregullt | `np.linspace(a,b,n)` | `linspace(a,b,n)` |
| Shumëzim elementar | `a*b` | `a.*b` |
| Shumëzim matricor | `A@B` | `A*B` |
| Transpozim pa konjugim | `A.T` | `A.'` |
| Numri i elementeve | `a.size` | `numel(a)` |
| Dimensionet | `a.shape` | `size(a)` |

Për vektorë ose matrica komplekse, `A'` kryen transpozimin e konjuguar; `A.'` vetëm transpozimin.


````octave
t = linspace(0.0, 5.0, 6);
zero = zeros(1, 4);
njesi = ones(1, 4);
A = [2.0, 1.0; 1.0, 3.0];
x = [1.0; 2.0];

disp(t);
disp(size(A));
fprintf('A(1,2) = %.1f (indeksimi fillon nga 1)\n', A(1,2));
disp('A*x ='); disp(A*x);
fprintf('Produkti skalar x''*x = %.1f\n', x'*x);
disp('Transpozimi:'); disp(A.');
fprintf('Norma e x = %.6f\n', norm(x));
fprintf('Mesatarja, minimumi, maksimumi: %.2f %.2f %.2f\n', mean(t), min(t), max(t));
````

### Vektorizimi dhe koha e ekzekutimit

Koha varet nga makina, versioni dhe bibliotekat. Eksperimenti nuk provon një ligj universal të performancës.


````octave
N = 200000;
x_test = linspace(0.0, 10.0, N);

y_loop = zeros(size(x_test));
tic;
for i = 1:N
    y_loop(i) = sin(x_test(i))^2 + cos(x_test(i))^2;
end
koha_loop = toc;

tic;
y_vec = sin(x_test).^2 + cos(x_test).^2;
koha_vec = toc;

fprintf('Cikli: %.4f s\n', koha_loop);
fprintf('Vektorizimi: %.4f s\n', koha_vec);
fprintf('Diferenca maksimale: %.3e\n', max(abs(y_loop-y_vec)));
````

## 5. Zbatimi kryesor: lëvizja me nxitim konstant

\[
x(t)=x_0+v_0t+\frac12at^2,\qquad v(t)=v_0+at.
\]

Ky është kampionim i zgjidhjes analitike në një rrjet diskret.


````octave
x0 = 1.0; v0 = 2.0; a = 1.5;
t = linspace(0.0, 6.0, 301);
x_pos = x0 + v0.*t + 0.5.*a.*t.^2;
v = v0 + a.*t;
assert(abs(x_pos(1)-x0) < 1e-12);
assert(abs(v(1)-v0) < 1e-12);
fprintf('x(6 s) = %.3f m\n', x_pos(end));
fprintf('v(6 s) = %.3f m/s\n', v(end));
````

````octave
figure;
subplot(2,1,1);
plot(t, x_pos, 'b-', 'LineWidth', 1.6);
ylabel('x (m)'); grid on; legend('Pozicioni');
subplot(2,1,2);
plot(t, v, 'r-', 'LineWidth', 1.6);
xlabel('t (s)'); ylabel('v (m/s)'); grid on; legend('Shpejtesia');
print('-dpng', '-r180', fullfile('figures', 'pozicioni_dhe_shpejtesia.png'));
````

````octave
figure; hold on;
for a_test = [-1.5, 0.0, 1.5, 3.0]
    x_test = x0 + v0.*t + 0.5.*a_test.*t.^2;
    plot(t, x_test, 'LineWidth', 1.4, 'DisplayName', sprintf('a=%.1f m/s^2', a_test));
end
xlabel('t (s)'); ylabel('x (m)'); title('Ndikimi i nxitimit');
grid on; legend('show'); hold off;
print('-dpng', '-r180', fullfile('figures', 'studim_parametrik_nxitimi.png'));
````

## 6. Lëvizja vertikale

\[
y(t)=y_0+v_0t-\frac12gt^2,\quad v(t)=v_0-gt,
\]

\[
t_{\rm imp}=\frac{v_0+\sqrt{v_0^2+2gy_0}}{g}.
\]


````octave
g = 9.81; y0 = 20.0; v0_y = 8.0; dt = 0.10;
t_vert = 0.0:dt:5.0;
y_vert = y0 + v0_y.*t_vert - 0.5.*g.*t_vert.^2;
v_vert = v0_y - g.*t_vert;
i_imp = find(y_vert <= 0.0, 1, 'first');
if isempty(i_imp)
    error('Intervali kohor nuk arrin deri te goditja.');
end
t_imp_diskret = t_vert(i_imp);
t_imp_sakte = (v0_y + sqrt(v0_y^2 + 2*g*y0))/g;
fprintf('Koha diskrete: %.3f s\n', t_imp_diskret);
fprintf('Koha analitike: %.6f s\n', t_imp_sakte);
````

````octave
figure;
subplot(2,1,1);
plot(t_vert, y_vert, 'g-', 'LineWidth', 1.5); hold on;
plot([t_imp_diskret t_imp_diskret], ylim, 'r--');
plot(xlim, [0 0], 'k-'); hold off;
ylabel('Lartesia y (m)'); grid on;
subplot(2,1,2);
plot(t_vert, v_vert, 'm-', 'LineWidth', 1.5); hold on;
plot([t_imp_diskret t_imp_diskret], ylim, 'r--'); hold off;
xlabel('Koha t (s)'); ylabel('Shpejtesia v (m/s)'); grid on;
print('-dpng', '-r180', fullfile('figures', 'levizja_vertikale.png'));
````

````octave
for dt_test = [0.40, 0.17, 0.05, 0.007]
    tt = 0.0:dt_test:5.0;
    yy = y0 + v0_y.*tt - 0.5.*g.*tt.^2;
    ii = find(yy <= 0.0, 1, 'first');
    t_num = tt(ii);
    fprintf('dt=%6.3f s -> t_imp=%.4f s, |gabimi|=%.4f s\n', dt_test, t_num, abs(t_num-t_imp_sakte));
end
````

## 9. Sfida opsionale

### Nxitimi pjesë-pjesë


````octave
t_piece = linspace(0.0, 6.0, 301); t_nd = 2.0;
x0_p = 0.0; v0_p = 1.0; a1 = 2.0; a2 = -1.0;
x_nd = x0_p + v0_p*t_nd + 0.5*a1*t_nd^2;
v_nd = v0_p + a1*t_nd;
x_piece = zeros(size(t_piece)); v_piece = zeros(size(t_piece));
mask1 = t_piece <= t_nd; mask2 = ~mask1;
x_piece(mask1) = x0_p + v0_p.*t_piece(mask1) + 0.5.*a1.*t_piece(mask1).^2;
v_piece(mask1) = v0_p + a1.*t_piece(mask1);
tau = t_piece(mask2)-t_nd;
x_piece(mask2) = x_nd + v_nd.*tau + 0.5.*a2.*tau.^2;
v_piece(mask2) = v_nd + a2.*tau;
````

### Hedhja dypërmasore


````octave
v0_proj = 20.0; theta = 40.0*pi/180.0;
t_flight = 2*v0_proj*sin(theta)/g;
t_proj = linspace(0.0, t_flight, 250);
x_proj = v0_proj*cos(theta).*t_proj;
y_proj = v0_proj*sin(theta).*t_proj - 0.5*g.*t_proj.^2;
figure; plot(x_proj, y_proj, 'b-', 'LineWidth', 1.5);
xlabel('Largesia horizontale x (m)'); ylabel('Lartesia y (m)');
title('Trajektorja analitike e predhes'); grid on;
print('-dpng', '-r180', fullfile('figures', 'sfida_hedhja_dypermasore.png'));
````
