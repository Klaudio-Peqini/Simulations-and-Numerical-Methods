### Vektorizimi

Vektorizimi zëvendëson ciklin eksplicit me një operacion mbi të gjithë vargun. Ai e afron kodin me shprehjen matematikore dhe shpesh përfiton nga bibliotekat e optimizuara. Kjo nuk do të thotë se çdo kod i vektorizuar është gjithmonë më i shpejtë.


````python
N = 200_000
x_test = np.linspace(0.0, 10.0, N)

fillimi = perf_counter()
y_loop = np.empty_like(x_test)
for i in range(N):
    y_loop[i] = np.sin(x_test[i])**2 + np.cos(x_test[i])**2
koha_loop = perf_counter() - fillimi

fillimi = perf_counter()
y_vec = np.sin(x_test)**2 + np.cos(x_test)**2
koha_vec = perf_counter() - fillimi

print(f"Cikli: {koha_loop:.4f} s")
print(f"Vektorizimi: {koha_vec:.4f} s")
print("Diferenca maksimale:", np.max(np.abs(y_loop-y_vec)))
````

> **Rezultati i pritshëm:** identiteti trigonometrik jep vlera shumë pranë 1 dhe diferenca midis dy zbatimeve duhet të jetë zero brenda saktësisë numerike. Koha varet nga makina dhe nuk duhet interpretuar si konstante universale.

## 5. Zbatimi kryesor: lëvizja me nxitim konstant

Modeli analitik është

$$
x(t)=x_0+v_0t+\frac12at^2, \qquad v(t)=v_0+at.
$$

Ne e vlerësojmë zgjidhjen analitike në një rrjet diskret kohor; **nuk** po integrojmë numerikisht një ekuacion diferencial.


````python
def levizja_nxitim_konstant(t_s, x0_m, v0_m_s, a_m_s2):
    # Kthen pozicionin dhe shpejtësinë për nxitim konstant.
    t_s = np.asarray(t_s, dtype=float)
    x_m = x0_m + v0_m_s*t_s + 0.5*a_m_s2*t_s**2
    v_m_s = v0_m_s + a_m_s2*t_s
    return x_m, v_m_s

x0, v0, a = 1.0, 2.0, 1.5
t = np.linspace(0.0, 6.0, 301)
x, v = levizja_nxitim_konstant(t, x0, v0, a)

assert np.isclose(x[0], x0)
assert np.isclose(v[0], v0)
print(f"x(6 s) = {x[-1]:.3f} m")
print(f"v(6 s) = {v[-1]:.3f} m/s")
````

````python
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(7.2, 7.0), sharex=True)
ax1.plot(t, x, color="darkblue", label="Pozicioni")
ax1.set_ylabel("$x$ (m)")
ax1.grid(True, alpha=0.3)
ax1.legend()

ax2.plot(t, v, color="darkred", label="Shpejtësia")
ax2.set_xlabel("$t$ (s)")
ax2.set_ylabel("$v$ (m/s)")
ax2.grid(True, alpha=0.3)
ax2.legend()

fig.suptitle("Lëvizja me nxitim konstant")
fig.tight_layout()
fig.savefig(FIG_DIR / "pozicioni_dhe_shpejtesia.png", dpi=180)
plt.show()
````

### Rastet kufitare dhe studimi parametrik

- (a=0): lëvizje e njëtrajtshme;
- (v_0=0): nisje nga prehja;
- (a<0): shpejtësia zvogëlohet dhe mund të ndryshojë shenjë.


````python
fig, ax = plt.subplots(figsize=(7.2, 4.5))
for a_test in [-1.5, 0.0, 1.5, 3.0]:
    x_test, _ = levizja_nxitim_konstant(t, x0, v0, a_test)
    ax.plot(t, x_test, label=fr"$a={a_test:.1f}$ m/s$^2$")
ax.set_xlabel("$t$ (s)")
ax.set_ylabel("$x$ (m)")
ax.set_title("Ndikimi i nxitimit")
ax.grid(True, alpha=0.3)
ax.legend()
fig.tight_layout()
fig.savefig(FIG_DIR / "studim_parametrik_nxitimi.png", dpi=180)
plt.show()
````

## 6. Shembull i integruar: lëvizja vertikale

Për drejtimin pozitiv lart,

$$
y(t)=y_0+v_0t-\frac12gt^2,\qquad v(t)=v_0-gt.
$$

Koha analitike pozitive e goditjes është

$$
t_{\rm imp}=\frac{v_0+\sqrt{v_0^2+2gy_0}}{g}.
$$


````python
g = 9.81
y0 = 20.0
v0_y = 8.0
dt = 0.10
t_vert = np.arange(0.0, 5.0 + dt, dt)
y_vert = y0 + v0_y*t_vert - 0.5*g*t_vert**2
v_vert = v0_y - g*t_vert

indekset = np.flatnonzero(y_vert <= 0.0)
if indekset.size == 0:
    raise RuntimeError("Intervali kohor nuk arrin deri te goditja.")
i_imp = indekset[0]
t_imp_diskret = t_vert[i_imp]
t_imp_sakte = (v0_y + np.sqrt(v0_y**2 + 2*g*y0))/g

print(f"Koha diskrete e goditjes: {t_imp_diskret:.3f} s")
print(f"Koha analitike: {t_imp_sakte:.6f} s")
print(f"Diferenca: {abs(t_imp_diskret-t_imp_sakte):.6f} s")
````

````python
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(7.2, 7.0), sharex=True)
ax1.plot(t_vert, y_vert, color="darkgreen")
ax1.axhline(0.0, color="black", linewidth=1)
ax1.axvline(t_imp_diskret, color="crimson", linestyle="--", label="Goditja diskrete")
ax1.set_ylabel("Lartësia $y$ (m)")
ax1.grid(True, alpha=0.3)
ax1.legend()

ax2.plot(t_vert, v_vert, color="purple")
ax2.axvline(t_imp_diskret, color="crimson", linestyle="--")
ax2.set_xlabel("Koha $t$ (s)")
ax2.set_ylabel("Shpejtësia $v$ (m/s)")
ax2.grid(True, alpha=0.3)
fig.tight_layout()
fig.savefig(FIG_DIR / "levizja_vertikale.png", dpi=180)
plt.show()
````

````python
for dt_test in [0.40, 0.17, 0.05, 0.007]:
    tt = np.arange(0.0, 5.0 + dt_test, dt_test)
    yy = y0 + v0_y*tt - 0.5*g*tt**2
    t_num = tt[np.flatnonzero(yy <= 0.0)[0]]
    print(f"dt={dt_test:6.3f} s -> t_imp={t_num:.4f} s, |gabimi|={abs(t_num-t_imp_sakte):.4f} s")
````

> **Interpretim fizik:** rrjeti kohor regjistron vetëm çaste të caktuara. Koha e parë me (y\leq0) zakonisht është pak pas kohës së saktë. Zvogëlimi i (\Delta t) e përmirëson lokalizimin, por ky është ende kampionim i një zgjidhjeje analitike, jo metodë Euler.

## 9. Sfida opsionale

### Sfida A — Nxitim konstant pjesë-pjesë
Përdorni (a_1=2\,\mathrm{m/s^2}) për (t<2\,\mathrm{s}) dhe (a_2=-1\,\mathrm{m/s^2}) më pas, duke ruajtur vazhdimësinë e (x) dhe (v).


````python
t_piece = np.linspace(0.0, 6.0, 301)
t_nd = 2.0
x0_p, v0_p, a1, a2 = 0.0, 1.0, 2.0, -1.0
x_nd = x0_p + v0_p*t_nd + 0.5*a1*t_nd**2
v_nd = v0_p + a1*t_nd
x_piece = np.where(
    t_piece <= t_nd,
    x0_p + v0_p*t_piece + 0.5*a1*t_piece**2,
    x_nd + v_nd*(t_piece-t_nd) + 0.5*a2*(t_piece-t_nd)**2,
)
v_piece = np.where(t_piece <= t_nd, v0_p+a1*t_piece, v_nd+a2*(t_piece-t_nd))
print("Vazhdimësia pranë ndarjes:", x_nd, v_nd)
````

### Sfida B — Hedhja dypërmasore
Për (v_0=20\,\mathrm{m/s}) dhe këndin (40^\circ), përdorni formulat analitike pa rezistencë ajri.


````python
v0_proj = 20.0
theta = np.deg2rad(40.0)
t_flight = 2*v0_proj*np.sin(theta)/g
t_proj = np.linspace(0.0, t_flight, 250)
x_proj = v0_proj*np.cos(theta)*t_proj
y_proj = v0_proj*np.sin(theta)*t_proj - 0.5*g*t_proj**2

fig, ax = plt.subplots(figsize=(7.0, 4.2))
ax.plot(x_proj, y_proj)
ax.set_xlabel("Largësia horizontale $x$ (m)")
ax.set_ylabel("Lartësia $y$ (m)")
ax.set_title("Trajektorja analitike e predhës")
ax.grid(True, alpha=0.3)
fig.tight_layout()
fig.savefig(FIG_DIR / "sfida_hedhja_dypermasore.png", dpi=180)
plt.show()
````
