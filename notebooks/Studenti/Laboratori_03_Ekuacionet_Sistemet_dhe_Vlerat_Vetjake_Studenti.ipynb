## Përgatitja teorike


Për $f(x)=0$, përgjysmimi garanton tkurrjen e intervalit kur $f(a)f(b)<0$, ndërsa Njutoni përdor linearizimin

$$
x_{k+1}=x_k-\frac{f(x_k)}{f'(x_k)}.
$$

Për $\mathbf{F}(\mathbf{x})=\mathbf{0}$, korrigjimi merret nga $J(\mathbf{x}_k)\Delta\mathbf{x}_k=-\mathbf{F}(\mathbf{x}_k)$. Në çdo hap zgjidhet një sistem linear; nuk formohet $J^{-1}$. Për $A\mathbf{v}=\lambda\mathbf{v}$, metoda fuqi ndërton $\mathbf{v}_{k+1}=A\mathbf{v}_k/\|A\mathbf{v}_k\|$ dhe konvergon drejt vektorit të vlerës vetjake me modul më të madh kur ekziston një hendek spektral.


## Algoritmi dhe kriteret e verifikimit


Përdorim një ekuacion të rënies me rezistencë për rrënjën skalare, kryqëzimin e rrethit me një drejtëz për sistemin jolinear dhe një matricë simetrike masë--sustë për modet normale. Çdo rezultat kontrollohet me mbetje dhe me një zgjidhje reference.
