## Përgatitja teorike


Nga zhvillimi Taylor merret formula qendrore
$$
f'(x)=\frac{f(x+h)-f(x-h)}{2h}+\mathcal{O}(h^2).
$$
Gabimi i këputjes ulet si $h^2$, ndërsa gabimi i rrumbullakimit rritet afërsisht si $u/h$. Për integrimin, trapezi i përbërë ka gabim global $\mathcal{O}(h^2)$ dhe Simpsoni $\mathcal{O}(h^4)$ për funksione mjaft të lëmuara. Kuadratura Gauss--Legendre me $n$ pika është e saktë për polinome deri në gradën $2n-1$.
