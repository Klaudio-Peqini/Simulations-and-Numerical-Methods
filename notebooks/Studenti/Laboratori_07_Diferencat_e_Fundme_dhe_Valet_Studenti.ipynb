## Përgatitja teorike


Për ekuacionin $u_{tt}=c^2u_{xx}$, skema qendrore është
$$
u_j^{n+1}=2u_j^n-u_j^{n-1}+\lambda^2(u_{j+1}^n-2u_j^n+u_{j-1}^n),
\qquad \lambda=\frac{c\Delta t}{\Delta x}.
$$
Në një përmasë kërkohet $\lambda\le1$. Për kufi të fiksuar vendoset $u=0$; shenja e pulsit përmbyset gjatë reflektimit. Dispersimi numerik do të thotë se shpejtësia fazore varet nga numri valor edhe kur modeli i vazhdueshëm nuk është dispersiv.
