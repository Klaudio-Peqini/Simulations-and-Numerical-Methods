## Përgatitja teorike


Një gjenerator pseudorastësor është determinist; fara përcakton vargun. Për funksion kumulativ të kthyeshëm, nëse $U\sim\mathcal U(0,1)$, atëherë $X=F^{-1}(U)$ ka shpërndarjen e synuar. Box--Muller përdor
$$
Z_1=\sqrt{-2\ln U_1}\cos(2\pi U_2),\qquad Z_2=\sqrt{-2\ln U_1}\sin(2\pi U_2).
$$
Në kampionimin me refuzim kërkohet $f(x)\le Mq(x)$; norma mesatare e pranimit është $1/M$ kur dendësitë janë të normalizuara.


## Algoritmi dhe kriteret e verifikimit

Kontrollojmë riprodhueshmërinë, çiftet e njëpasnjëshme, transformimin eksponencial, Box--Muller-in dhe refuzimin për një dendësi trekëndore.
