## Përgatitja teorike


Për shpërndarjen objektiv $\pi(x)$ dhe propozimin $q(y|x)$, pranimi është
$$
\alpha(x,y)=\min\left[1,\frac{\pi(y)q(x|y)}{\pi(x)q(y|x)}\right].
$$
Në modelin Ising 2D, $E=-J\sum_{\langle ij\rangle}s_is_j$ dhe $M=\sum_i s_i$. Për përmbysjen e një spini,
$$
\Delta E=2Js_{ij}\sum_{nn}s_{nn}.
$$
Pas termalizimit maten $e=E/N$ dhe $m=M/N$. Nxehtësia specifike merret nga luhatjet, $C_V=(\langle E^2\rangle-\langle E\rangle^2)/(NT^2)$. Matjet e njëpasnjëshme janë të korreluara; gabimi duhet vlerësuar me bllokim ose autokorrelacion.

```python
def energjia(s,J=1.0): return -J*np.sum(s*(np.roll(s,1,0)+np.roll(s,1,1)))

def sweep(s,beta,rng,J=1.0):
    # TODO: plotësoni zbatimin duke u mbështetur te algoritmi i mësipërm.
    raise NotImplementedError('Plotësoni këtë pjesë')

def simulim(T,L=24,burn=500,measure=1200,thin=2):
    s=rng.choice([-1,1],(L,L))
    for _ in range(burn): sweep(s,1/T,rng)
    E=[]; M=[]
    for k in range(measure*thin):
        sweep(s,1/T,rng)
        if k%thin==0: E.append(energjia(s)); M.append(s.sum())
    return s,np.array(E),np.array(M)
```

### Eksperimenti Python 3


```python
Ts=np.linspace(1.5,3.5,14); mag=[]; cv=[]
for T in Ts:
    s,E,M=simulim(T); N=s.size; mag.append(np.mean(np.abs(M))/N); cv.append(np.var(E,ddof=1)/(N*T*T))
fig,axs=plt.subplots(1,2,figsize=(10,4)); axs[0].plot(Ts,mag,'o-'); axs[1].plot(Ts,cv,'o-');
axs[0].set(xlabel='T',ylabel='$\\langle|m|\\rangle$'); axs[1].set(xlabel='T',ylabel='$C_V$'); plt.show()
```
